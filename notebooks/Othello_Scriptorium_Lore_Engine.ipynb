# Othello — Scriptorium Lore Engine

Procedural companion notebook for the **Othello Liturgical Commune**.
Use this notebook in Google Colab to:
1. Generate canonical `LC-XXX` IDs and map entries to the Rule of Hours.
2. Combine liturgical motifs, choir vows, and antiphon structures into fresh lore seeds.
3. Optionally draft Synod commentary or canticles with Gemini (`google-genai`).

```python
import random
from datetime import datetime, timezone

LITURGICAL_CATEGORIES = [
    "Canticle_Or_Hymn",
    "Canon_Precept",
    "Rite_Or_Sacrament",
    "Hagiography_Or_Martyr",
    "Relic_Or_Vestment",
    "Commune_Chronicle",
    "Apocrypha_Or_Heresy",
]

CANONICAL_HOURS = [
    ("Vigils", "Before dawn; silence broken only by the bronze stalls"),
    ("Lauds", "First light; ink grinding and morning chronicle reading"),
    ("Prime", "Working hour; allotment of tools, reed pens, and labor"),
    ("Sext", "Midday; shared salt, cold water, and wax-tablet grievances"),
    ("Vespers", "Dusk; bell-tolling and Synod reading of grievances"),
    ("Compline", "Nightfall; sealing of the reliquary and lamps dimmed"),
]

CHOIRS_AND_HOUSES = [
    "Choir of the First Watch",
    "House of the Common Table",
    "Scriptorium of the Ash-Ink",
    "Order of Bell-Founders",
    "Hermits of the Sealed Marginalia",
]

def next_lore_id(existing_ids):
    nums = [int(i.split("-")[1]) for i in existing_ids if i.startswith("LC-")]
    return f"LC-{(max(nums, default=0) + 1):03d}"

print("Sample next ID:", next_lore_id(["LC-001", "LC-002", "LC-003"]))
```

```python
MOTIFS = [
    "a cracked bronze censer that hums before winter rain",
    "an iron salt-ledger whose pages were scraped clean and rewritten",
    "a bell-rope braided from the linen shrouds of former precentors",
    "a glossed psalter whose margin notes contradict the chanted text",
    "a sunken cistern beneath the refectory used for silent penance",
]

ANTIPHON_FRAMES = [
    "Keep the {noun} unbroken, lest the outer basin forget its name.",
    "No scribe sits higher than the {noun} at the hour of {hour}.",
    "What is sealed in {noun} returns to voice when the Synod sleeps.",
]

def generate_lore_seed(seed=None):
    rng = random.Random(seed)
    hour, hour_note = rng.choice(CANONICAL_HOURS)
    category = rng.choice(LITURGICAL_CATEGORIES)
    house = rng.choice(CHOIRS_AND_HOUSES)
    motif = rng.choice(MOTIFS)
    noun = rng.choice(["wick", "ash-ink", "salt-bowl", "bell-clapper", "vesper-lamp"])
    antiphon = rng.choice(ANTIPHON_FRAMES).format(noun=noun, hour=hour)
    return {
        "submitted_at": datetime.now(timezone.utc).isoformat(),
        "liturgical_category": category,
        "canonical_hour_or_season": hour,
        "hour_observance": hour_note,
        "commune_house_or_choir": house,
        "epigraph_or_antiphon": antiphon,
        "central_motif": motif,
    }

for idx in range(2):
    print(f"Seed {idx+1}:", generate_lore_seed(42 + idx))
```
